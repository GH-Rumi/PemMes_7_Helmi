```python
corpus = [
    'the house had a tiny little mouse',
    'the cat saw the mouse',
    'the mouse ran away from the house',
    'the cat finally ate the mouse',
    'the end of the mouse story'
]
```

Pada langkah ini, Anda menyiapkan data mentah berupa sebuah daftar (list) bernama corpus. Daftar ini berisi lima kalimat singkat berbahasa Inggris yang akan bertindak sebagai kumpulan dokumen (korpus) untuk dianalisis pada tahap selanjutnya.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Inisiasi obyek TFidfVectorizer
vect = TfidfVectorizer(stop_words='english')

# Pembobotan TF-IDF
resp = vect.fit_transform(corpus)

# Cetak hasil
print(resp)
```

Di tahap ini, Anda menginisiasi algoritma TfidfVectorizer dan menginstruksikannya untuk secara otomatis membuang kata-kata umum bahasa Inggris yang tidak memiliki makna signifikan (stop words, seperti 'the', 'a', 'of', dsb.). Melalui perintah fit_transform, model menghitung bobot tiap kata yang tersisa. Output matriks yang dicetak menunjukkan lokasi kata dalam format (indeks dokumen, indeks kata) yang diikuti oleh angka desimal yang merepresentasikan bobot pentingnya kata tersebut.

```python
print(vect.get_feature_names_out())
```

Langkah terakhir ini memanggil fungsi get_feature_names_out() untuk menampilkan kamus atau daftar kata unik (vocabulary) yang berhasil diekstrak dari seluruh dokumen. Daftar kata ini diurutkan secara alfabetis, dan posisi kata pada daftar inilah yang menjadi acuan indeks kolom pada matriks angka di Langkah 1 (misalnya, angka 0 mewakili kata 'ate', angka 2 mewakili kata 'cat', dan angka 7 mewakili kata 'mouse').
