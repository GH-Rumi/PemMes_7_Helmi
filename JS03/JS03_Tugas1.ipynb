# **1. Import Library**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report

# Load data
df = pd.read_csv("wbc.csv")
```

# **2. Pembersihan Data**

```python
data_tidak_digunakan = ['id', 'Unnamed: 32']
df_bersih = df.drop(columns=[col for col in data_tidak_digunakan if col in df.columns])
df_bersih.head(10)
```

Membuat daftar kolom yang tidak memiliki nilai medis atau prediktif, yaitu id dan Unnamed: 32 (kolom kosong bawaan dari file CSV). Fungsi .drop() membuang kolom-kolom tersebut sehingga data menjadi lebih bersih sebelum diproses.

# **3. Encoding Target**

```python
le = LabelEncoder()
df_bersih['diagnosis'] = le.fit_transform(df_bersih['diagnosis'])
df_bersih.head(10)
```

Algoritma ML hanya dapat memproses angka. Oleh karena itu, LabelEncoder digunakan untuk mengubah nilai teks pada kolom diagnosis (M untuk Malignant dan B untuk Benign) menjadi angka biner (misalnya 1 dan 0).

# **4. Standardisasi Skala Fitur**

```python
scaler = StandardScaler()

X = df_bersih.drop(columns=['diagnosis'])
y = df_bersih['diagnosis']

X_scaled = scaler.fit_transform(X)

X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

print(X_scaled_df.head())
```

Data dibagi menjadi variabel target (y) dan variabel fitur prediksi (X). Karena fitur-fitur seperti radius dan area memiliki rentang satuan yang jauh berbeda, StandardScaler digunakan untuk menyamakan skala semua fitur numerik agar tidak ada variabel yang mendominasi model. Hasilnya kemudian dikembalikan ke dalam format tabel (Dataframe).

# **5. Pembagian Data, Pipeline, dan Evaluasi**

```python
# Membagi data menjadi data latih (80%) dan uji (20%) menggunakan data yang sudah distandardisasi
X_train, X_test, y_train, y_test = train_test_split(X_scaled_df, y, test_size=0.2, random_state=42, stratify=y)

# Buat Pipeline untuk Seleksi Fitur (SelectKBest) dan Model (Logistic Regression)
pipeline = Pipeline([
    ('selector', SelectKBest(score_func=f_classif, k=19)), # Menggunakan k=19 berdasarkan jumlah fitur yang paling optimal
    ('classifier', LogisticRegression(max_iter=1000, random_state=42))
])

# Latih model menggunakan data latih
pipeline.fit(X_train, y_train)

# Lakukan prediksi pada data uji
y_pred = pipeline.predict(X_test)

# Cetak hasil evaluasi model
print("=== Hasil Pengujian Model ===")
print("Akurasi:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

- train_test_split: Membagi data menjadi 80% untuk melatih model dan 20% untuk menguji model. Parameter stratify=y memastikan proporsi kelas diagnosis seimbang di kedua porsi data.

- Pipeline: Menggabungkan proses seleksi 19 fitur terbaik (SelectKBest) dan algoritma klasifikasi (LogisticRegression) dalam satu alur kerja yang rapi.

- fit & predict: Melatih model dengan data latih, lalu memintanya menebak diagnosis pada data uji. Hasil tebakan dievaluasi menggunakan metrik akurasi dan classification report.

# **6. Inspeksi Fitur Terpilih**

```python
# Ambil status fitur (True/False) dari SelectKBest
mask = pipeline.named_steps['selector'].get_support()

# Cocokkan dengan nama kolom dari dataframe X
fitur_terpilih = X_scaled_df.columns[mask].tolist()

print(f"Jumlah fitur yang digunakan: {len(fitur_terpilih)}")
print("Daftar Fitur Terbaik Terpilih:")
for i, fitur in enumerate(fitur_terpilih, start=1):
    print(f"{i}. {fitur}")
```

Fungsi get_support() dari pipeline mengambil daftar boolean (True/False) yang menandakan fitur mana saja dari 30 fitur awal yang lolos seleksi SelectKBest. Masker True/False ini dicocokkan dengan nama kolom asli pada data X untuk mencetak 19 nama fitur yang paling krusial dalam mendiagnosis kanker.
