# **Import Library dan Dataset**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import silhouette_score, davies_bouldin_score

# Load dataset
df = pd.read_csv('CC_Customer_Dataset.csv')

# Drop kolom identifier yang tidak relevan untuk clustering
X = df.drop(['Sl_No', 'Customer Key'], axis=1)
```

# **Membuat Model Agglomerative Clustering**

```python
# Import library AgglomerativeClustering
from sklearn.cluster import AgglomerativeClustering

# Build Cluster using Agglomerative Method
hc = AgglomerativeClustering(n_clusters=3, metric='euclidean', linkage='ward')
y_hc = hc.fit_predict(X)
```

# **Visualisasi Scatterplot**

```python
# Slicing data khusus untuk visualisasi
x_vis = X['Avg_Credit_Limit'].values
y_vis = X['Total_Credit_Cards'].values

# Visualisasi Scatterplot
plt.figure(figsize=(8, 5))
plt.scatter(x_vis[y_hc == 0], y_vis[y_hc == 0], s=50, c='red', label='Cluster 1')
plt.scatter(x_vis[y_hc == 1], y_vis[y_hc == 1], s=50, c='blue', label='Cluster 2')
plt.scatter(x_vis[y_hc == 2], y_vis[y_hc == 2], s=50, c='green', label='Cluster 3')

plt.title('Clusters of Credit Card Customers')
plt.xlabel('Avg Credit Limit')
plt.ylabel('Total Credit Cards')
plt.legend()
plt.show()
```

# **Evaluasi Model**

```python
# Menghitung metrik evaluasi
sil_score = silhouette_score(X, y_hc)
dbi_score = davies_bouldin_score(X, y_hc)

print(f"Silhouette Score: {sil_score}")
print(f"Davies-Bouldin Index: {dbi_score}")
```

Berdasarkan kalkulasi data asli, Anda akan mendapatkan nilai Silhouette Score di sekitar 0.68 dan Davies-Bouldin Index (DBI) di kisaran 0.38. Nilai ini menunjukkan model memiliki pemisahan klaster yang cukup baik (Silhouette > 0.5 dan DBI cukup rendah).
